# Ploting Univariate Numerical

```python
# line plot

plt.plot(df["close"], color="#01050FF0", linestyle="-", marker="*")
plt.grid()
plt.show()
```

```python
# Histogram:

plt.hist(df["close"], bins=5, color="#4A8AB5B6")
plt.show()
```
